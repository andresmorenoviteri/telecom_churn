```python
binary_cols = ["ChildrenInHH", "HandsetRefurbished", "HandsetWebCapable", "TruckOwner", "RVOwner", "Homeownership", "BuysViaMailOrder", "RespondsToMailOffers", "OptOutMailings", "NonUSTravel", "OwnsComputer", "HasCreditCard", "NewCellphoneUser", "NotNewCellphoneUser", "OwnsMotorcycle", "MadeCallToRetentionTeam"]

generic_map = [
    {"Yes": 1, "No": 0},
    {"Known": 1, "Unknown": 0}
]

for col in binary_cols:
    for m in generic_map:
        if set(df_kmedoids[col].dropna().unique()) == set(m.keys()):
            df_kmedoids[col] = df_kmedoids[col].map(m)
            break
```

```python
# create a flag for the missing value in "HandsetPrice"
df_kmedoids["HandsetPrice_missing"] = (df_kmedoids["HandsetPrice"] == "Unknown").astype(int)
df_kmedoids["HandsetPrice"] = pd.to_numeric(df_kmedoids["HandsetPrice"], errors="coerce")

median = df_kmedoids["HandsetPrice"].median()
df_kmedoids["HandsetPrice"] = df_kmedoids["HandsetPrice"].fillna(median)

df_kmedoids.head()
```

```python
def combine_status(row):
    if row["NewCellphoneUser"] == 1:
        return "New"
    elif row["NotNewCellphoneUser"] == 1:
        return "NotNew"
    else:
        return "Unknown"

df_kmedoids["CellphoneUserStatus"] = df_kmedoids.apply(combine_status, axis=1)
df_kmedoids = df_kmedoids.drop(columns=["NewCellphoneUser", "NotNewCellphoneUser"])
df_kmedoids["CellphoneUserStatus"] = df_kmedoids["CellphoneUserStatus"].astype("category")

df_kmedoids
```

```python
# non binary features, change to one hot rod
new_feature_cols = ["MaritalStatus", "CreditRating", "PrizmCode", "Occupation", "CellphoneUserStatus"]
for col in new_feature_cols:
    df_kmedoids = pd.get_dummies(df_kmedoids, columns=[col], drop_first=True, dtype=int)

df_kmedoids
```

```python
# scale the continuous value columns

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler

continues_cols = ['MonthlyRevenue', 'MonthlyMinutes',
       'TotalRecurringCharge', 'DirectorAssistedCalls', 'OverageMinutes',
       'RoamingCalls', 'PercChangeMinutes', 'PercChangeRevenues',
       'DroppedCalls', 'BlockedCalls', 'UnansweredCalls', 'CustomerCareCalls',
       'ThreewayCalls', 'ReceivedCalls', 'OutboundCalls', 'InboundCalls',
       'PeakCallsInOut', 'OffPeakCallsInOut', 'DroppedBlockedCalls',
       'CallForwardingCalls', 'CallWaitingCalls', 'MonthsInService',
       'UniqueSubs', 'ActiveSubs', 'Handsets', 'HandsetModels',
       'CurrentEquipmentDays', 'AgeHH1', 'AgeHH2', 'HandsetPrice']

ct = ColumnTransformer(
    [("num", RobustScaler(), continues_cols)],
    remainder="passthrough",
    verbose_feature_names_out=False
).set_output(transform="pandas")

df_kmedoids_scaled = ct.fit_transform(df_kmedoids)

df_kmedoids_scaled
```

```python
import kmedoids
from sklearn.metrics import silhouette_score, pairwise_distances
import numpy as np

# convert scaled dataframe to plain float array
x_train = df_kmedoids_scaled.to_numpy(dtype=float)
assert not np.isnan(x_train).any(), "There are NaN values in the input data"

# create a seeded random generator
rng = np.random.default_rng(seed=42)
# picks up to 10,000 random row indices without repeats
idx = rng.choice(x_train.shape[0], size=min(10000, x_train.shape[0]), replace=False)
# xsub is the subset
xsub = x_train[idx]

# D is a 10k by 10k matix of manhattan disntaces between all subsample rows
D = pairwise_distances(xsub, metric='manhattan')
km = kmedoids.fasterpam(D, 2, random_state=42)
medoids = xsub[km.medoids]

train_clusters = pairwise_distances(x_train, medoids, metric='manhattan').argmin(axis=1)

print(np.bincount(train_clusters))
print(silhouette_score(x_train, train_clusters, metric='manhattan'))

print(pd.Series(y_train.values).groupby(train_clusters).mean())
```

```python
X = df.drop(columns=["Churn", "CustomerID"])
y = df["Churn"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)
```

```python
x_train.head()
```

```python
x_train["MaritalStatus"].unique()
```

```python
x_train.drop(columns=["ServiceArea", "PrizmCode", "CreditRating"], inplace=True)
```

```python
generic_map = [
    {"Yes": 1, "No": 0, "Unknown": 0.5},
    {"Known": 1, "Unknown": 0},
    {'Other': 0, 'Clerical': 1, 'Professional': 2, 'Self': 3, 'Crafts': 4, 'Student': 5, 'Retired': 6, 'Homemaker': 7},
    {"New": 1, "Unknown": 0.5, "NotNew": 0}
]

cols = ["Occupation", "MaritalStatus", "CellphoneUserStatus"]

for col in cols:
    vals = set(x_train[col].dropna().unique())
    for m in generic_map:
        if vals <= set(m.keys()):          # subset instead of ==
            x_train[col] = x_train[col].map(m)
            break
    else:
        print(f"No map matched for {col}: {vals}")
```

```python
x_train.head()
```

```python
from sklearn_extra.cluster import KMedoids 
from sklearn.metrics import silhouette_score

kmedoids = KMedoids(n_clusters=2, random_state=42, metric="manhattan")  # metric, not metrics
kmedoids.fit(x_train)

labels = kmedoids.labels_
medoid_idx = kmedoids.medoid_indices_      # indices (note: medoid_indices_, not medoids_indices_)
medoids = kmedoids.cluster_centers_        # the actual medoid points

silhouette_avg = silhouette_score(
    x_train, labels,
    metric="manhattan",                    # match the clustering metric
    sample_size=10000, random_state=42     # silhouette is O(n²), so subsample on big data
)
print(silhouette_avg)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report, roc_auc_score, confusion_matrix, RocCurveDisplay
)


xgb_model = XGBClassifier(
    n_estimators=300, max_depth=5, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8,
    eval_metric="logloss", enable_categorical=True, random_state=42
)
xgb_model.fit(x_train, y_train)


# predictions
y_pred = xgb_model.predict(x_test)
y_proba = xgb_model.predict_proba(x_test)[:, 1]


# evaluation
print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```
