```python
# for binary columns with missing values, convert to string category including "Missing"
cols_binary_with_na = [col for col in binary_feat_col if df[col].isna().sum() > 0]

for col in cols_binary_with_na:
    df[col] = df[col].fillna("missing").astype(str)

# pass these columns to categorical features
categorical_feat_col += cols_binary_with_na
binary_feat_col = [col for col in binary_feat_col if col not in cols_binary_with_na]

# for categorical columns (including the ones just moved) with missing values, fill with "missing"
cols_categorical_with_na = [col for col in categorical_feat_col if df[col].isna().sum() > 0]

for col in cols_categorical_with_na:
    df[col] = df[col].fillna("missing").astype(str)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

# get X and y from dataset
X = df.drop(columns=["churn", "Customer_ID"])
y = df["churn"]

# split the data to create features
x_train, x_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

# lets create a flag column for each column that has a null value 
# then get median of column for these values
imputer = SimpleImputer(strategy="median", add_indicator=True)
x_train_imputed_numeric = imputer.fit_transform(x_train[numerical_feat_col])
x_test_imputed_numeric = imputer.transform(x_test[numerical_feat_col])
```

```python
# create the names for the missing columns
missing_cols = [numerical_feat_col[i] + "_missing" for i in imputer.indicator_.features_]
new_columns = numerical_feat_col + missing_cols

# switch arrays back to dataframes with column names
x_train_imputed_numeric_df = pd.DataFrame(x_train_imputed_numeric, columns=new_columns, index=x_train.index)
x_test_imputed_numeric_df = pd.DataFrame(x_test_imputed_numeric, columns=new_columns, index=x_test.index)
```

```python
x_train_df = pd.concat([x_train[binary_feat_col + categorical_feat_col], x_train_imputed_numeric_df], axis=1)
x_test_df = pd.concat([x_test[binary_feat_col + categorical_feat_col], x_test_imputed_numeric_df], axis=1)
```

```python
binary_cols = []
for col in x_train_df.columns:
    if x_train_df[col].nunique() == 2:
        binary_cols.append(col)
```

```python
# do one hot encoder

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore", drop="first", sparse_output=False)
x_train_cat = encoder.fit_transform(x_train_df[updated_categorical_feat_col])
x_test_cat = encoder.transform(x_test_df[updated_categorical_feat_col])

cat_columns = encoder.get_feature_names_out(updated_categorical_feat_col)
x_train_cat_df = pd.DataFrame(x_train_cat, columns=cat_columns, index=x_train_df.index)
x_test_cat_df = pd.DataFrame(x_test_cat, columns=cat_columns, index=x_test_df.index)

# drop original categorical columns, add encoded ones
x_train_final = pd.concat([x_train_df.drop(columns=updated_categorical_feat_col), x_train_cat_df], axis=1)
x_test_final = pd.concat([x_test_df.drop(columns=updated_categorical_feat_col), x_test_cat_df], axis=1)
```

### Raw data trained on XGBoost

```python
df.head()
```

```python
X = df.drop(columns=["churn", "Customer_ID"])
y = df["churn"]
```

```python
x_train_raw, x_test_raw, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report, roc_auc_score, confusion_matrix, RocCurveDisplay
)

cat_cols = ["new_cell", "crclscod", "asl_flag", "prizm_social_one", "area",
            "dualband", "refurb_new", "hnd_webcap", "ownrent", "dwlltype",
            "marital", "infobase", "HHstatin", "dwllsize", "ethnic",
            "kid0_2", "kid3_5", "kid6_10", "kid11_15", "kid16_17", "creditcd"]

for col in cat_cols:
    x_train_raw[col] = x_train_raw[col].astype("category")
    x_test_raw[col] = x_test_raw[col].astype("category")

xgb_model = XGBClassifier(
    n_estimators=300, max_depth=5, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8,
    eval_metric="logloss", enable_categorical=True, random_state=42
)
xgb_model.fit(x_train_raw, y_train)


# predictions
y_pred = xgb_model.predict(x_test_raw)
y_proba = xgb_model.predict_proba(x_test_raw)[:, 1]


# evaluation
print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
RocCurveDisplay.from_predictions(y_test, y_proba)
plt.show()

# feature importance
importance_df = pd.DataFrame({
    "feature": x_train_final.columns,
    "importance": xgb_model.feature_importances_
}).sort_values("importance", ascending=False)

print(importance_df.head(20))

plt.figure(figsize=(8, 8))
top20 = importance_df.head(20).sort_values("importance")
plt.barh(top20["feature"], top20["importance"])
plt.title("Top 20 XGBoost Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```
